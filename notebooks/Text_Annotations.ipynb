```python
import os
from pathlib import Path
import pandas as pd
from PIL import Image
import matplotlib.pyplot as plt
```

```python
# Define your root directory based on the image provided
root_dir = Path("../../data/CelebAMask-HQ")
img_dir = root_dir / "CelebA-HQ-img"
mask_dir = root_dir / "CelebAMask-HQ-mask-anno"

# Load attributes into a DataFrame for exploration
attr_path = root_dir / "CelebAMask-HQ-attribute-anno.txt"
```

```python
# Note: First line is count, second is header
df_attr = pd.read_csv(attr_path, sep='\s+', skiprows=1)
print(df_attr.columns)
print(f"\nNumber of features: {len(df_attr.columns)}")
```

```python
import matplotlib.pyplot as plt
from PIL import Image
from pathlib import Path

def print_single_sample(n):
    """
    Displays the i-th image and prints its attribute description.
    Assumes img_dir and df_attr are defined in the global scope.
    """
    # 1. Path Construction
    for i in range(n):
        img_path = img_dir / f"{i}.jpg"
            
        # 2. Image Loading
        img = Image.open(img_path)
        
        # 3. Visualization
        plt.figure(figsize=(8, 8))
        plt.imshow(img)
        plt.title(f"CelebAMask-HQ Index: {i}")
        plt.axis('off')
        plt.show()
    
        # 4. Description Extraction and Printout
        # Extract columns where the value is 1 (presence of attribute)
        sample_row = df_attr.iloc[i]
        active_attributes = sample_row[sample_row == 1].index.tolist()
        
        # Clean output formatting
        description = ", ".join(active_attributes)
        print(f"--- Description for Sample {i} ---")
        print(description)
        print("-" * 30)

# Example: Print the 5th image (index 4)
print_single_sample(10)
```
